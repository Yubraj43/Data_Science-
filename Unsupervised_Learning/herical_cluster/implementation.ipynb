# Hierarchical Clustering Implementation\n
Basic agglomerative clustering example.

```python
import numpy as np\n
import matplotlib.pyplot as plt\n
from sklearn.datasets import make_blobs\n
from sklearn.cluster import AgglomerativeClustering
```

```python
# Sample data (replace with your own dataset if needed)\n
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=1.0, random_state=42)
```

```python
# Build hierarchical clustering model\n
model = AgglomerativeClustering(n_clusters=4, linkage='ward')\n
labels = model.fit_predict(X)
```

```python
# Visualize clusters\n
plt.figure(figsize=(7, 5))\n
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=30)\n
plt.title('Hierarchical Clustering Result')\n
plt.xlabel('Feature 1')\n
plt.ylabel('Feature 2')\n
plt.grid(alpha=0.3)\n
plt.show()
```
