# PCA Implementation\n
Principal Component Analysis with scikit-learn.

```python
import numpy as np\n
import pandas as pd\n
import matplotlib.pyplot as plt\n
from sklearn.preprocessing import StandardScaler\n
from sklearn.decomposition import PCA
```

```python
# Replace with your dataset\n
# df = pd.read_csv('your_dataset.csv')\n
# X = df.select_dtypes(include=['number'])
```

```python
# Example demo data (remove this block if you load your own dataset)\n
rng = np.random.RandomState(42)\n
X = rng.randn(200, 5)
```

```python
# 1) Standardize\n
scaler = StandardScaler()\n
X_scaled = scaler.fit_transform(X)
```

```python
# 2) Fit PCA\n
pca = PCA(n_components=2, random_state=42)\n
X_pca = pca.fit_transform(X_scaled)\n
print('Transformed shape:', X_pca.shape)
```

```python
# 3) Explained variance\n
print('Explained variance ratio:', pca.explained_variance_ratio_)\n
print('Cumulative explained variance:', pca.explained_variance_ratio_.cumsum())
```

```python
# 4) 2D visualization\n
plt.figure(figsize=(7, 5))\n
plt.scatter(X_pca[:, 0], X_pca[:, 1], alpha=0.7)\n
plt.title('PCA Projection (2 Components)')\n
plt.xlabel('PC1')\n
plt.ylabel('PC2')\n
plt.grid(True, alpha=0.3)\n
plt.show()
```
