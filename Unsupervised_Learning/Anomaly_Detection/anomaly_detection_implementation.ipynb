# Anomaly Detection Implementation
Using Isolation Forest on synthetic 2D data.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.ensemble import IsolationForest
```

```python
# Generate normal data
X_normal, _ = make_blobs(n_samples=300, centers=1, cluster_std=0.9, random_state=42)

# Generate synthetic anomalies
rng = np.random.RandomState(42)
X_anomaly = rng.uniform(low=-8, high=8, size=(30, 2))

# Combine data
X = np.vstack([X_normal, X_anomaly])
```

```python
# Train Isolation Forest
model = IsolationForest(contamination=0.09, random_state=42)
pred = model.fit_predict(X)

# In sklearn: -1 = anomaly, 1 = normal
anomaly_mask = pred == -1

print('Total points:', len(X))
print('Detected anomalies:', anomaly_mask.sum())
```

```python
# Visualization
plt.figure(figsize=(8, 6))
plt.scatter(X[~anomaly_mask, 0], X[~anomaly_mask, 1], c='steelblue', s=25, label='Normal')
plt.scatter(X[anomaly_mask, 0], X[anomaly_mask, 1], c='crimson', s=40, label='Anomaly')
plt.title('Anomaly Detection using Isolation Forest')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
