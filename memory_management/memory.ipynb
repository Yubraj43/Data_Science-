```python
#merory management in python
import sys
import gc
class myobject:
    def __init__(self, name):
        self.name = name
    def __del__(self):
        print(f"{self.name} is being deleted")
obj1 = myobject("object1")
print(sys.getrefcount(obj1))
obj2 = obj1
print(sys.getrefcount(obj1))
del obj1
print(sys.getrefcount(obj2))
del obj2
gc.collect()
```

```python
#generation garbage collection
import gc
def generate_numbers(n):
    for i in range(n):
        yield i
gen = generate_numbers(1000000)
print(gc.get_count())
gc.collect()
print(gc.get_count())
```
