# Naive Bayes Implementation

This notebook demonstrates:
1. Gaussian Naive Bayes on numeric data (Iris)
2. Multinomial Naive Bayes on text spam classification

## 1) Imports

```python
from sklearn.datasets import load_iris
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, MultinomialNB
```

## 2) Gaussian Naive Bayes (Numeric Data)

```python
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data,
    iris.target,
    test_size=0.2,
    random_state=42,
    stratify=iris.target
)

gaussian_model = GaussianNB()
gaussian_model.fit(X_train, y_train)
gaussian_pred = gaussian_model.predict(X_test)

print('Gaussian Naive Bayes - Iris Dataset')
print('Accuracy:', round(accuracy_score(y_test, gaussian_pred), 4))
print(classification_report(y_test, gaussian_pred, target_names=iris.target_names))
```

## 3) Multinomial Naive Bayes (Text Spam Classification)

```python
messages = [
    'Win a free iPhone now',
    'Congratulations you won a lottery claim now',
    'Urgent claim your cash prize',
    'Hi are you coming to class today',
    'Let us meet for lunch tomorrow',
    'Can you call me when you are free',
    'Free reward waiting click now',
    'Project meeting is scheduled for tomorrow'
]
labels = [1, 1, 1, 0, 0, 0, 1, 0]  # 1 = spam, 0 = not spam

X_train_msg, X_test_msg, y_train_msg, y_test_msg = train_test_split(
    messages,
    labels,
    test_size=0.25,
    random_state=42,
    stratify=labels
)

vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train_msg)
X_test_vec = vectorizer.transform(X_test_msg)

multinomial_model = MultinomialNB(alpha=1.0)
multinomial_model.fit(X_train_vec, y_train_msg)
multinomial_pred = multinomial_model.predict(X_test_vec)

print('Multinomial Naive Bayes - Simple Spam Dataset')
print('Accuracy:', round(accuracy_score(y_test_msg, multinomial_pred), 4))
print(classification_report(y_test_msg, multinomial_pred, target_names=['not spam', 'spam']))
```

## 4) Test on New Messages

```python
sample_texts = [
    'Claim your free reward now',
    'Let us discuss the assignment tomorrow'
]

sample_vec = vectorizer.transform(sample_texts)
sample_pred = multinomial_model.predict(sample_vec)

for text, pred in zip(sample_texts, sample_pred):
    label = 'spam' if pred == 1 else 'not spam'
    print(f"{text!r} -> {label}")
```
