## simple linear regression 
simple linear regression is mmechine learning algorithm used to predict a continuous value using one input variables 
# 1 is input (x)
# 2 is output (y)

# y= predicted value 
# x=input variable
# b0=intercept
# b1=slope

# formula of simple linear regression :y=b0+b1x

```python
#import liberaries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#Create a simple dataset
data={'Hours':[1,2,3,4,5],
      'Scores':[10,20,30,40,50]}
df=pd.DataFrame(data)
print(df)
```

```python
#define x and y
x=df[['Hours']]
y=df['Scores']
```

```python
#train model
model = LinearRegression()
model.fit(x,y)
```

```python
model.predict([[6]])
```

```python
#plot Regression line 
plt.scatter(x,y,color='blue')
plt.plot(x,model.predict(x),color='red')
plt.xlabel('Hours')
plt.ylabel('Scores')
plt.show()
```


# why linear regression is importance

demand prediction


```python
from sklearn.metrics import r2_score
r2_score(y,model.predict(x))
```

# multi linear regression 


```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    "Size":[1000,1500,1800],
    "Rooms":[2,3,3],
    "Price":[200000,300000,350000]
}

df = pd.DataFrame(data)

X = df[["Size","Rooms"]]
y = df["Price"]

model = LinearRegression()
model.fit(X,y)

model.predict([[1600,3]])
```

```python
plt.scatter(df["Size"],df["Price"],color='blue')
plt.scatter(df["Rooms"],df["Price"],color='green')
plt.xlabel('Size/Rooms')
plt.ylabel('Price')
plt.show()
```
