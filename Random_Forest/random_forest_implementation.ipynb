# Random Forest Implementation (Classifier + Regressor)

This notebook includes:
1) RandomForestClassifier on Iris dataset
2) RandomForestRegressor on synthetic regression data

## 1) Imports

```python
from sklearn.datasets import load_iris, make_regression
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.metrics import accuracy_score, classification_report, mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
```

## 2) RandomForestClassifier (Iris)

```python
iris = load_iris()
X_cls = iris.data
y_cls = iris.target

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_cls, y_cls, test_size=0.2, random_state=42, stratify=y_cls
)

rf_clf = RandomForestClassifier(
    n_estimators=200,
    max_depth=6,
    random_state=42
)

rf_clf.fit(X_train_c, y_train_c)
y_pred_c = rf_clf.predict(X_test_c)

print('RandomForestClassifier Results')
print('Accuracy:', round(accuracy_score(y_test_c, y_pred_c), 4))
print(classification_report(y_test_c, y_pred_c, target_names=iris.target_names))
```

```python
print('Feature Importances (Classifier):')
for name, score in zip(iris.feature_names, rf_clf.feature_importances_):
    print(f'{name}: {score:.4f}')
```

## 3) RandomForestRegressor (Synthetic Data)

```python
X_reg, y_reg = make_regression(n_samples=400, n_features=6, noise=20, random_state=42)

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

rf_reg = RandomForestRegressor(
    n_estimators=250,
    max_depth=8,
    random_state=42
)

rf_reg.fit(X_train_r, y_train_r)
y_pred_r = rf_reg.predict(X_test_r)

mae = mean_absolute_error(y_test_r, y_pred_r)
mse = mean_squared_error(y_test_r, y_pred_r)
r2 = r2_score(y_test_r, y_pred_r)

print('RandomForestRegressor Results')
print('MAE:', round(mae, 4))
print('MSE:', round(mse, 4))
print('R2 Score:', round(r2, 4))
```

```python
print('Feature Importances (Regressor):')
for idx, score in enumerate(rf_reg.feature_importances_):
    print(f'feature_{idx}: {score:.4f}')
```
