Python logging is the standard way to print useful messages from your program (info, warnings, errors) in a controlled, configurable way. It’s better than print() because you can change level, format, and where logs go (console/file) without changing your code everywhere.

```python
import logging
```

```python
#confingure logging
logging.basicConfig(level=logging.DEBUG)
```

```python
#log some messages
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

```python
#configure logging
logging.basicConfig(filename='app.log', level=logging.DEBUG)
#log some messages
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

```python
#configure logging with format and filename
logging.basicConfig(filename='lo1.log', level=logging.DEBUG, format='%(asctime)s - %(levelname)s - %(message)s')
#log some messages
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

```python
#configure logging with format datefmt and filename as well as level
logging.basicConfig(filename='lo2.log', level=logging.DEBUG, format='%(asctime)s - %(levelname)s - %(message)s', datefmt='%Y-%m-%d %H:%M:%S')
#log some messages
logging.debug("This is a debug message")    
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
    
```
