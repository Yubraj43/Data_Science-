“Multi logger” means using more than one logger (or more than one output) in the same program.
1.mutiple logger (different name)
Example
      one for the database 
      another for the API
      

```python
import logging
```

```python
#create a logger for module1\
logger1 = logging.getLogger('module1')
logger1.setLevel(logging.DEBUG)
#create a logger for module2
logger2 = logging.getLogger('module2')
logger2.setLevel(logging.INFO)
#configure logging for module1
handler1 = logging.FileHandler('module1.log')
formatter1 = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
handler1.setFormatter(formatter1)
logger1.addHandler(handler1)

#configure logging for module2
handler2 = logging.FileHandler('module2.log')
formatter2 = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
handler2.setFormatter(formatter2)
logger2.addHandler(handler2)
#log some messages
logger1.debug("This is a debug message from module1")
logger1.info("This is an info message from module1")
logger1.warning("This is a warning message from module1")
logger1.error("This is an error message from module1")
logger1.critical("This is a critical message from module1")
logger2.debug("This is a debug message from module2")
logger2.info("This is an info message from module2")
logger2.warning("This is a warning message from module2")
logger2.error("This is an error message from module2")
logger2.critical("This is a critical message from module2")
```
